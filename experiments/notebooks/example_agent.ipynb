# Orion AGI Agent Example

This notebook demonstrates how to create and use an AGI agent with the Orion framework.

```python
import sys
sys.path.insert(0, '../..')

from src.agents.agi_agent import AGIAgent
from src.core.base import Goal
from datetime import datetime
import asyncio
```

## 1. Create an AGI Agent

```python
agent = AGIAgent(
    agent_id="notebook_agent_1",
    name="Notebook Agent",
    capabilities=["reasoning", "learning", "planning"],
    config={
        "short_term_capacity": 10,
        "working_capacity": 7
    }
)

print(f"Created agent: {agent.name}")
print(f"Capabilities: {agent.capabilities}")
```

## 2. Set a Goal

```python
goal = Goal(
    id="goal_1",
    description="Learn to navigate a grid environment",
    priority=1,
    status="active",
    created_at=datetime.now()
)

agent.add_goal(goal)
agent.current_goal = goal

print(f"Goal set: {goal.description}")
```

## 3. Run Agent Cycles

```python
async def run_agent_cycles(num_cycles=3):
    results = []
    
    for i in range(num_cycles):
        environment = {
            "position": [i, i],
            "objects": ["wall", "door", "key"],
            "goal_position": [5, 5]
        }
        
        result = await agent.run_cycle(environment)
        results.append(result)
        
        print(f"\nCycle {result['cycle']}:")
        print(f"  Learning complete: {result['learning_complete']}")
    
    return results

results = await run_agent_cycles(3)
```

## 4. Check Agent Status

```python
status = agent.get_status()

print("Agent Status:")
print(f"  Agent ID: {status['agent_id']}")
print(f"  Name: {status['name']}")
print(f"  Interactions: {status['interaction_count']}")
print(f"\nMemory Statistics:")
for key, value in status['memory_statistics'].items():
    print(f"  {key}: {value}")
print(f"\nLearning Evaluation:")
for key, value in status['learning_evaluation'].items():
    print(f"  {key}: {value}")
```

## 5. Explore Memory System

```python
# Get context from working memory
context = agent.memory_system.get_context(limit=5)

print(f"Working Memory Context ({len(context)} items):")
for i, memory in enumerate(context):
    print(f"\nMemory {i+1}:")
    print(f"  Type: {memory.memory_type}")
    print(f"  Importance: {memory.importance}")
    print(f"  Access count: {memory.access_count}")
```

## 6. Visualize Learning Progress

```python
import matplotlib.pyplot as plt

learning_history = agent.learner.get_learning_history(limit=10)

if learning_history:
    cycles = [i for i in range(len(learning_history))]
    
    plt.figure(figsize=(10, 6))
    plt.plot(cycles, [1] * len(cycles), 'o-')
    plt.xlabel('Cycle')
    plt.ylabel('Learning Events')
    plt.title('Agent Learning Progress')
    plt.grid(True)
    plt.show()
else:
    print("No learning history available yet")
```
